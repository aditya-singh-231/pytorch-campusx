In this I have used 'nn' module for creating the model, used built in activation functions, used built in loss funtions, and used built in optimizers, unlike the previous model made where everything is coded and no modules or inbuilt functions, optimizers are used.

```python
#import the reqd. libraries
import torch
import torch.nn as nn

# creating a model class


"""
#normal code

class Model(nn.Module):
  def __init__(self, num_features): # num_features will tell the no. of fearures in our input
    super().__init__()
    self.linear=nn.Linear(num_features, 1) #we have to get only 1 output
    self.sigmoid=nn.Sigmoid()

  def forward(self, features):   # features are the data that we are inputting in the model
    out=self.linear(features)    #here we are calculating 'z=wx+b', here 'out' is our 'z'
    out=self.sigmoid(out)
    return out

"""


"""
#improved code 1 (using hidden layers)

# now we are building a model with hidden layers
class Model(nn.Module):
  # in constructor we will define the structure of our neural network
  def __init__(self, num_features): # num_features will tell the no. of fearures in our input
    super().__init__()
    self.linear1=nn.Linear(num_features, 3)
    self.relu=nn.ReLU()
    self.linear2=nn.Linear(3,1)
    self.sigmoid=nn.Sigmoid()

  def forward(self, features):

    out=self.linear1(features)
    out=self.relu(out)
    out=self.linear2(out)
    out=self.sigmoid(out)
    return out

  """

  # improved code 2 (using containers (Sequential))

class Model(nn.Module):
    # in constructor we will define the structure of our neural network
  def __init__(self, num_features): # num_features will tell the no. of fearures in our input
    super().__init__()
    self.network=nn.Sequential(
      nn.Linear(num_features,3),
      nn.ReLU(),
      nn.Linear(3,1),
      nn.Sigmoid()
    )

  def forward(self, features):
    out=self.network(features)
    return out
```

```python
#create dataset
features=torch.rand(10,5) # a dataset with 10 rows and 5 columns--- 5 columns because we have 5 features

#create a model
model=Model(features.shape[1])  # the value of features.shape[1] is 5

#call model for forward pass
model(features)  # we can also do 'model.forward(features)', but the prev. one is recommended by'PyTorch'

# whenever we call the object of the 'Model' class, then automatically 'forward' method of this class is called. PyTorch has overwritten the '__call__' method in 'nn.Module' class for this thing.
```

in the above o/p we got, we did not did backpropagation, all we did was sending the rows and getting the corresponding y_pred
